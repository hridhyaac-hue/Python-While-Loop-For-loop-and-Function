## Number Guessing Game

```python
import random
secret_number = random.randint(1,10)
print("welcome to the number guessing game")
total_attempts = 3
while total_attempts>0:
    guess = int(input("Guess a number between 1 and 10:"))  
    if guess <1 or guess >10:
        print("Invalid number, number must be in between 1 and 10")
        continue
    total_attempts = total_attempts-1
    if guess > secret_number:
        print("Too High, Try Again")
    elif guess < secret_number:
        print("Too Low, Try Again")
    else:
        print("congrats, Your Guess is correct")
        break
else:
    print("Better Luck next time!")
    print("The Secret Number is:",secret_number)
```

## Multiplication Table Generator

```python
# Problem Statement: Create a Python program that generates and prints a multiplication table (from 1 to 10) for a given number using a for loop and the range function
```

```python
number = int(input("Enter the number for which you want the multiplication table:"))
for i in range(1,11):
    product = number * i
    print(number,"x",i,"=",product)
```

## BMI Calculator

```python
# Problem Statement: Create a Python program that calculates the Body Mass Index(BMI).
# Hint: BMI = weight (kg) / [height (m)]²
```

```python
def BMI_calculator(weight,height):
    BMI = weight / (height**2)
    return BMI
weight = float(input("Enter your weight in kg:"))
height = float(input("Enter your height in meters:"))
BMI = BMI_calculator(weight,height)
print("Your BMI is:",round(BMI,2))
```
